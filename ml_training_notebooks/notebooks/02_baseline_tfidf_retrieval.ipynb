# Baseline TF-IDF Retrieval

Sanity check retrieval di atas dataset prepared hasil parsing dari raw CSV.


```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

from src.config import INTERIM_DIR
from src.preprocessing import load_prepared_dataset
from src.retrieval import build_tfidf_retriever, retrieve_top_k
```

```python
df = load_prepared_dataset(INTERIM_DIR)
vectorizer, matrix = build_tfidf_retriever(df['perfume_text'].fillna(''))
print('Loaded rows:', len(df))
print('Unique brands:', df['brand'].nunique(dropna=True))
print('Unique countries:', df['country'].nunique(dropna=True))
print('TF-IDF matrix shape:', matrix.shape)
```

```python
query = 'warm woody amber vanilla date night'
results = retrieve_top_k(df, query, vectorizer, matrix, top_k=10)
display(results[['perfume_name', 'brand', 'country', 'gender', 'year', 'accords', 'baseline_score']])
```

```python
queries = [
    'fresh citrus aromatic office fragrance',
    'rose oud sweet evening perfume',
    'powdery floral elegant feminine scent',
]

for query in queries:
    print(f'\nQuery: {query}')
    sample = retrieve_top_k(df, query, vectorizer, matrix, top_k=5)
    display(sample[['perfume_name', 'brand', 'country', 'baseline_score']])
```
